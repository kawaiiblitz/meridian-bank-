# Development branch off main — creation captured in code (executed)

Creates the `development` branch **off `main`** (git) and off the clean branch (Lakebase), with the commands and their output.

```python
import subprocess
def sh(c):
    print('$ '+c)
    print(subprocess.run(c,shell=True,capture_output=True,text=True,cwd='.').stdout or subprocess.run(c,shell=True,capture_output=True,text=True,cwd='.').stderr)
print('ready')
```

```python
# git: development branch OFF main
sh("git branch development main   # creates development off main")
sh("git merge-base main development   # shared root => development came off main")
sh("git log --oneline --graph main development | head -12")
```

```python
# Lakebase: development branch off the clean branch
sh("databricks postgres get-branch projects/meridian-bank/branches/development -o json")
```
